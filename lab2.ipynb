```python
from random import random, sample, seed, randint
from icecream import ic
from collections.abc import Sequence
from math import inf
```

```python
N = 10
seed(42)
```

```python
OBJECTS = frozenset(range(N))
SETS = tuple(frozenset(sample(range(N), k=s+1)) for s in range(N))
COSTS = tuple(10*random()+(s+random())**2 for s in range(N))
```

## ARRAY OF BOOLS

```python
def random_solution() -> list[bool]: # creates a random solution
    return sample([True, False], k=N, counts=(N, N))

def fitness_aob(solution: Sequence[bool]) -> tuple[int, float]:

    # fitness = (number of covered objects, -weight)
    # covering more objects is better, if the number of covered objects is equal, a lower cost is better
    # when comparing tuples, the first element is compared first, and if they are equal, the second element is compared

    covered = frozenset().union(*(SETS[n] for n in range(N) if solution[n]))
    weight = sum(COSTS[n] for n in range(N) if solution[n])
    return len(covered), -weight

def is_valid(solution: Sequence[bool]) -> bool:

    # solution valid if it covers all objects

    return fitness_aob(solution)[0] == len(OBJECTS)

def tweak(solution: Sequence[bool]) -> list[bool]:

    # flipping a random index in the solution to create a new solution

    new_solution = list(solution)  # copy
    i = randint(0, N - 1)  # random index to flip
    new_solution[i] = not new_solution[i] # flip
    return new_solution

def hill_climbing(solution: list[bool], max_no_improvement: int = 50):

    # stopping criterion -> steady state -> if no improvement is found after max_no_improvement iterations, stop

    fit = fitness_aob(solution)
    no_improvement = 0
    while no_improvement < max_no_improvement:
        candidate = tweak(solution)
        candidate_fit = fitness_aob(candidate)
        if candidate_fit > fit:
            solution, fit = candidate, candidate_fit
            no_improvement = 0
        else:
            no_improvement += 1
    return solution, fit


def hill_climbing_restarts(restarts: int = 20):
    best, best_fit = None, (-1, -inf)
    for _ in range(restarts):
        sol, fit = hill_climbing(random_solution())
        if fit > best_fit:
            best, best_fit = sol, fit
    return best, best_fit
```

```python
hill_climbing_restarts()
```

```python
for N in range(1, 11): # create 10 different instances

    OBJECTS = frozenset(range(N))
    SETS = tuple(frozenset(sample(range(N), k=s+1)) for s in range(N))
    COSTS = tuple(10*random()+(s+random())**2 for s in range(N))

    best, best_fit = hill_climbing_restarts()

    ic(N)
    ic(SETS)
    ic(COSTS)
    ic(best, best_fit)
    ic("-----------------------------------")
```

```python
for N in range(1, 101): # create 100 different instances

    OBJECTS = frozenset(range(N))
    SETS = tuple(frozenset(sample(range(N), k=s+1)) for s in range(N))
    COSTS = tuple(10*random()+(s+random())**2 for s in range(N))

    best, best_fit = hill_climbing_restarts()

    ic(N)
    ic(best, best_fit)
```
