```python
from random import random, sample, seed, randint
from icecream import ic
from collections.abc import Sequence
from math import inf
```

```python
N = 7
#seed(42)
```

```python
OBJECTS = {n for n in range(N)} # crea un set di oggetti
SETS = tuple(frozenset(sample(range(N), k=s+1)) for s in range(N))
COSTS = tuple(10*random()+(s+random())**2 for s in range(N))
```

## ARRAY OF BOOLS

```python
solution = sample([True, False], k=N, counts=(N,N)) # crea una soluzione casuale
solution
```

```python
def fitness_aob(solution: Sequence[bool]) -> float:
    weight = sum(COSTS[n] for n in range(N) if solution[n])

    if solution != [False for _ in range(N)] and frozenset.union(*(SETS[n] for n in range(N) if solution[n])) == frozenset(OBJECTS):
        return -weight
    return -inf
    #check = frozenset.union(*(SETS[n] for n in range(N) if solution[n]))
    #assert check == frozenset(OBJECTS)
    #return -weight
```

```python
fitness_aob(solution)
```

```python
def tweak(solution: Sequence[bool]) -> Sequence[bool]:
    new_solution = solution.copy() # create a copy of the solution to modify
    i = randint(0, len(solution) - 1) # pick a random index to flip
    new_solution[i] = not new_solution[i] # flip the value at that index
    return new_solution
```

```python
solution, tweak(solution)
```

```python
steps = 0
no_improvement = 0
ic(steps, fitness_aob(solution))

while True:
    steps += 1
    new_solution = tweak(solution)
    ic(new_solution)
    if fitness_aob(new_solution) > fitness_aob(solution):
        no_improvement = 0
        solution = new_solution.copy()
        ic(steps, fitness_aob(solution))
    else:
        no_improvement += 1
    if no_improvement >= 50:  # Stop if no improvement for 50 iterations
        break

sets = [SETS[n] for n in range(N) if solution[n]]
ic(solution, fitness_aob(solution), sets)
```

## SET OF SETS

```python
XCOSTS = {SETS[i]: COSTS[i] for i in range(N)}
XCOSTS
```

```python
solution = set(sample(SETS, k=5))
solution
```

```python
def fitness_sos(solution: set[frozenset]) -> float:
    weight = sum(XCOSTS[s] for s in solution)
    if frozenset.union(*solution) == frozenset(OBJECTS):
        return -weight
    return 0.0
```
